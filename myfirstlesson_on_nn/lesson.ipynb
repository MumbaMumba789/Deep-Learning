```python
import torch

# The relationship we want the model to discover
weight = 0.7
bias = 0.3

# Create X values
X = torch.arange(0, 1, 0.02).unsqueeze(dim=1)

# Create the correct answers
y = weight * X + bias

print(X[:5])
print(y[:5])
```

```python
train_split = int(0.8 * len(X))

X_train = X[:train_split]
y_train = y[:train_split]

X_test = X[train_split:]
y_test = y[train_split:]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
import torch.nn as nn

class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()

        self.weight = nn.Parameter(torch.randn(1))
        self.bias = nn.Parameter(torch.randn(1))

    def forward(self, x):
        return self.weight * x + self.bias


torch.manual_seed(42)
model = LinearRegressionModel()

print(model.state_dict())
```

```python
# Make predictions using the training data
y_pred = model(X_train)

# Create a function to measure the error
loss_fn = nn.L1Loss()

# Compare predictions with the correct answers
loss = loss_fn(y_pred, y_train)

print("First predictions:", y_pred[:5])
print("Correct answers:", y_train[:5])
print("Initial loss:", loss.item())
```

```python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

# Calculate gradients from the loss
optimizer.zero_grad()
loss.backward()

# Update the model's weight and bias
optimizer.step()

print("Updated parameters:", model.state_dict())
```

```python
epochs = 1000
for epoch in range(epochs):
    #make the predictions
    y_pred = model(X_train)
    #calculate the loss
    loss = loss_fn(y_pred, y_train)
    #clear old gradients
    optimizer.zero_grad()
    #calculate new gradients
    loss.backward()
    #update the new gradients
    optimizer.step()

    if epoch % 10 ==0:
        print(f"Epoch: {epoch}, Loss: {loss.item():.4f}")
        
```

```python
# Put the model in evaluation mode
model.eval()

# Make predictions on data the model did not train on
with torch.inference_mode():
    test_preds = model(X_test)
    test_loss = loss_fn(test_preds, y_test)

# Compare learned values with the original values
print("Original weight:", weight)
print("Original bias:", bias)

print("Learned parameters:", model.state_dict())
print("Test loss:", test_loss.item())

print("Predictions:", test_preds[:5].squeeze())
print("Correct answers:", y_test[:5].squeeze())
```

```python
from pathlib import Path

MODEL_PATH = Path('models')
MODEL_PATH.mkdir(parents=True, exist_ok=True)
MODEL_NAME = "01_pytorch_workflow_model.pth"
MODEL_SAVE_PATH = MODEL_PATH / MODEL_NAME

print(f"Saving model to: {MODEL_SAVE_PATH}")
torch.save(obj=model.state_dict(),
           f = MODEL_SAVE_PATH)
```
